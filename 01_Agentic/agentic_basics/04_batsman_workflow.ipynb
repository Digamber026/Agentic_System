```python
from langgraph.graph import StateGraph, END, START
from typing import TypedDict
```

```python
class batsState(TypedDict):
    runs:int
    balls:int
    fours:int
    sixes:int

    sr:float
    b_percent : float
    bpb:float
    summary : str
```

```python
def calc_sr(state:batsState)-> batsState:
    runs = state['runs']
    balls = state['balls']
    sr = (runs/balls)*100
    return {'sr' :sr}
```

```python
def calc_b(state:batsState):
    six = state['sixes']
    four = state['fours']
    run = state['runs']
    bp = ((six*6 + four*4)/run)*100
    
    return {'b_percent':bp}
```

```python
def calc_bpb(state:batsState):
    bpb = state['balls']/(state['fours']+state['sixes'])
    return {'bpb' : bpb}
    
```

```python
def calc_summary(state:batsState):
    summary = f""" 
    Strike rate {state['sr']}\n
    Runs in boundary% {state['b_percent']}\n
    Balls per boundary {state['bpb']} """

    return {'summary': summary}
```

```python
graph = StateGraph(batsState)

graph.add_node('calc_sr', calc_sr)
graph.add_node('calc_b',calc_b)
graph.add_node('calc_bpb', calc_bpb)
graph.add_node('calc_summary', calc_summary)

graph.add_edge(START, 'calc_sr')
graph.add_edge(START, 'calc_b')
graph.add_edge(START, 'calc_bpb')

graph.add_edge('calc_sr' , 'calc_summary')
graph.add_edge('calc_b', 'calc_summary')
graph.add_edge('calc_bpb', 'calc_summary')
graph.add_edge('calc_summary',END)
```

```python
workflow = graph.compile()
```

```python
initial_state = {
    'runs' : 300,
    'balls': 100,
    'sixes':20,
    'fours':10
}

final_output = workflow.invoke(initial_state);
print(final_output['summary'])
#print(final_output)
```
